```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------
# 0. KONFIGURIMI KRYESOR
# ----------------------------------------------------

# NDRYSHO VETËM KËTË LINE sipas asaj që del në Colab pas unzip-it
DATA_DIR = "/content/recording/000"   # ose "/content/recording" nëse s'ka folder 000

CONF_THRESHOLD = 0.8     # minimum confidence
MIN_FIX_DUR = 80         # ms
MAX_FIX_DUR = 800        # ms


# ----------------------------------------------------
# 1. NGARKO TË DHËNAT NGA EKSPORTI
# ----------------------------------------------------

def load_exported_data(data_dir: str):
    gaze_path = os.path.join(data_dir, "gaze_positions.csv")
    fix_path = os.path.join(data_dir, "fixations.csv")
    blink_path = os.path.join(data_dir, "blinks.csv")

    gaze = pd.read_csv(gaze_path)
    fixations = pd.read_csv(fix_path)

    blinks = None
    if os.path.exists(blink_path):
        blinks = pd.read_csv(blink_path)

    # surfaces (opsionale)
    surf_files = glob.glob(os.path.join(data_dir, "surfaces", "surface_gaze_positions*.csv"))
    surface_gaze = pd.read_csv(surf_files[0]) if surf_files else None

    return gaze, fixations, blinks, surface_gaze


# ----------------------------------------------------
# 2. PASTRIMI I GAZE & FIXIMEVE
# ----------------------------------------------------

def clean_gaze(gaze: pd.DataFrame):
    df = gaze.copy()

    # timestamp
    if "timestamp" in df.columns:
        tcol = "timestamp"
    elif "world_timestamp" in df.columns:
        tcol = "world_timestamp"
    else:
        raise ValueError("Nuk gjeta kolonën timestamp te gaze_positions.csv")

    df = df.rename(columns={tcol: "timestamp"})

    # pozicioni 2D
    if "norm_pos_x" in df.columns and "norm_pos_y" in df.columns:
        pass
    elif "gaze_point_2d_x" in df.columns and "gaze_point_2d_y" in df.columns:
        df = df.rename(columns={
            "gaze_point_2d_x": "norm_pos_x",
            "gaze_point_2d_y": "norm_pos_y"
        })
    else:
        raise ValueError("Nuk gjeta kolonat e pozicionit të gaze")

    # filtro sipas confidence
    if "confidence" in df.columns:
        df = df[df["confidence"] >= CONF_THRESHOLD]

    df = df.dropna(subset=["norm_pos_x", "norm_pos_y", "timestamp"])
    return df.reset_index(drop=True)


def clean_fixations(fixations: pd.DataFrame):
    df = fixations.copy()

    # kolonë start_timestamp
    if "start_timestamp" in df.columns:
        start_col = "start_timestamp"
    elif "timestamp" in df.columns:
        start_col = "timestamp"
    else:
        raise ValueError("Nuk gjeta kolonën start_timestamp te fixations.csv")

    df = df.rename(columns={start_col: "start_timestamp"})

    if "duration" not in df.columns:
        raise ValueError("'duration' nuk ekziston te fixations.csv")

    df["duration_ms"] = df["duration"] * 1000.0
    df["end_timestamp"] = df["start_timestamp"] + df["duration"]

    df = df[(df["duration_ms"] >= MIN_FIX_DUR) & (df["duration_ms"] <= MAX_FIX_DUR)]
    return df.reset_index(drop=True)


def attach_blinks_to_gaze(gaze, blinks):
    df = gaze.copy()
    df["is_blink"] = False

    if blinks is None or blinks.empty:
        return df

    for _, row in blinks.iterrows():
        onset = row.get("onset", None)
        offset = row.get("offset", None)
        if onset is None or offset is None:
            continue

        mask = (df["timestamp"] >= onset) & (df["timestamp"] <= offset)
        df.loc[mask, "is_blink"] = True

    # heq kampionët gjatë blink
    df = df[~df["is_blink"]].reset_index(drop=True)
    return df
```

```python
# ----------------------------------------------------
# 3. BASHKIMI GAZE + FIXATIONS
# ----------------------------------------------------

def merge_gaze_and_fixations(gaze, fixations):
    g = gaze.copy().sort_values("timestamp").reset_index(drop=True)
    f = fixations.copy().sort_values("start_timestamp").reset_index(drop=True)

    f["fixation_id"] = np.arange(len(f))
    g["fixation_id"] = -1

    for _, fx in f.iterrows():
        mask = (g["timestamp"] >= fx["start_timestamp"]) & (g["timestamp"] <= fx["end_timestamp"])
        g.loc[mask, "fixation_id"] = fx["fixation_id"]

    # mbajmë vetëm gaze që i takojnë një fiksimi
    g = g[g["fixation_id"] >= 0].reset_index(drop=True)

    # shtojmë duration e fiksimit në çdo rresht
    g = g.merge(f[["fixation_id", "duration_ms"]], on="fixation_id", how="left")

    return g


# ----------------------------------------------------
# 4. PIPELINE PËR NJË SESION
# ----------------------------------------------------

def run_session(data_dir):
    gaze, fixations, blinks, surface_gaze = load_exported_data(data_dir)

    print("Gaze raw:", gaze.shape)
    print("Fixations raw:", fixations.shape)

    gaze_clean = clean_gaze(gaze)
    fix_clean = clean_fixations(fixations)
    gaze_clean = attach_blinks_to_gaze(gaze_clean, blinks)

    gaze_fix = merge_gaze_and_fixations(gaze_clean, fix_clean)

    print("\nPastruar:")
    print("Gaze:", gaze_clean.shape)
    print("Fixations:", fix_clean.shape)
    print("Merged gaze+fix:", gaze_fix.shape)

    # vizualizim i fiksimeve
    plt.hist(fix_clean["duration_ms"], bins=30)
    plt.xlabel("Fixation duration (ms)")
    plt.ylabel("Count")
    plt.title("Distribuimi i kohëzgjatjes së fiksimeve")
    plt.show()

    print("\nShembuj të të dhënave të bashkuara:")
    print(gaze_fix.head())

    return gaze_fix, fix_clean, surface_gaze


# EKZEKUTIMI
gaze_fix_df, fix_clean_df, surface_gaze_df = run_session(DATA_DIR)
```

```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------
# 0. KONFIGURIMI KRYESOR
# ----------------------------------------------------

DATA_DIR = "/content/recording/000"   # ndrysho nëse te printimi i mëparshëm s’ke folder 000

CONF_THRESHOLD = 0.8
MIN_FIX_DUR = 80
MAX_FIX_DUR = 800


# ----------------------------------------------------
# 1. NGARKO TË DHËNAT NGA EKSPORTI
# ----------------------------------------------------

def load_exported_data(data_dir: str):
    gaze_path = os.path.join(data_dir, "gaze_positions.csv")
    fix_path = os.path.join(data_dir, "fixations.csv")
    blink_path = os.path.join(data_dir, "blinks.csv")

    print("Lexoj prej:", gaze_path)
    gaze = pd.read_csv(gaze_path)
    fixations = pd.read_csv(fix_path)

    blinks = None
    if os.path.exists(blink_path):
        blinks = pd.read_csv(blink_path)

    surf_files = glob.glob(os.path.join(data_dir, "surfaces", "surface_gaze_positions*.csv"))
    surface_gaze = pd.read_csv(surf_files[0]) if surf_files else None

    print("Kolonat te gaze_positions.csv:", list(gaze.columns))
    print("Kolonat te fixations.csv:", list(fixations.columns))

    return gaze, fixations, blinks, surface_gaze


# ----------------------------------------------------
# 2. PASTRIMI I GAZE & FIXIMEVE
# ----------------------------------------------------

def clean_gaze(gaze: pd.DataFrame):
    df = gaze.copy()

    # ---------- GJEJMË KOLONËN E KOHËS ----------
    time_candidates = [c for c in df.columns if "timestamp" in c.lower() or c.lower() == "time"]
    if not time_candidates:
        raise ValueError(
            f"Nuk gjeta kolonë kohe te gaze_positions.csv. Kolonat aktuale janë: {list(df.columns)}"
        )
    tcol = time_candidates[0]
    df = df.rename(columns={tcol: "timestamp"})

    # ---------- GJEJMË KOLONAT E POZICIONIT 2D ----------
    x_col = None
    y_col = None

    # rastet më të zakonshme
    if "norm_pos_x" in df.columns and "norm_pos_y" in df.columns:
        x_col, y_col = "norm_pos_x", "norm_pos_y"
    elif "gaze_point_2d_x" in df.columns and "gaze_point_2d_y" in df.columns:
        x_col, y_col = "gaze_point_2d_x", "gaze_point_2d_y"
    else:
        # kërko çfarëdo kolone që përmban 'x' dhe 'y' në emër
        possible_x = [c for c in df.columns if ("x" in c.lower() and "norm" in c.lower()) or "gaze_point" in c.lower() and "x" in c.lower()]
        possible_y = [c for c in df.columns if ("y" in c.lower() and "norm" in c.lower()) or "gaze_point" in c.lower() and "y" in c.lower()]
        if possible_x and possible_y:
            x_col, y_col = possible_x[0], possible_y[0]

    if x_col is None or y_col is None:
        raise ValueError(
            f"Nuk gjeta kolonat e pozicionit 2D te gaze_positions.csv. Kolonat: {list(df.columns)}"
        )

    df = df.rename(columns={x_col: "norm_pos_x", y_col: "norm_pos_y"})

    # ---------- FILTROJMË SIPAS CONFIDENCE ----------
    if "confidence" in df.columns:
        df = df[df["confidence"] >= CONF_THRESHOLD]

    df = df.dropna(subset=["norm_pos_x", "norm_pos_y", "timestamp"])
    return df.reset_index(drop=True)


def clean_fixations(fixations: pd.DataFrame) -> pd.DataFrame:
    """
    Pastron fixations:
      - zgjedh kolonën e kohës së fillimit
      - detekton automatikisht njësi të 'duration'
      - llogarit duration_ms
      - heq fiksimet shumë të shkurtra/jashtë normales
    """
    df = fixations.copy()

    # kolonë fillimi – te ti është 'start_timestamp'
    if "start_timestamp" in df.columns:
        start_col = "start_timestamp"
    elif "timestamp" in df.columns:
        start_col = "timestamp"
    else:
        raise ValueError(f"Nuk gjeta kolonën e fillimit te fixations.csv. Kolonat: {list(df.columns)}")

    df = df.rename(columns={start_col: "start_timestamp"})

    if "duration" not in df.columns:
        raise ValueError(f"Nuk gjeta kolonën 'duration' te fixations.csv. Kolonat: {list(df.columns)}")

    dur = df["duration"].dropna()
    if dur.empty:
        raise ValueError("Kolona 'duration' është bosh.")

    med = dur.median()

    # vendosim njësi:
    # nëse median është mes 0.01–1 → sekonda
    # nëse median është mes 50–2000 → ms
    if 0.01 <= med <= 1.0:
        # duration në sekonda
        df["duration_ms"] = df["duration"] * 1000.0
        end_offset = df["duration"]          # sekonda
    else:
        # supozojmë që është në ms
        df["duration_ms"] = df["duration"]
        end_offset = df["duration_ms"] / 1000.0  # e kthejmë në sek për end_timestamp

    df["end_timestamp"] = df["start_timestamp"] + end_offset

    # filtro fiksimet (kufij pak më të gjerë)
    df = df[(df["duration_ms"] >= 40) & (df["duration_ms"] <= 1200)]

    return df.reset_index(drop=True)



def attach_blinks_to_gaze(gaze, blinks):
    df = gaze.copy()
    df["is_blink"] = False

    if blinks is None or blinks.empty:
        return df

    onset_cols = [c for c in blinks.columns if "onset" in c.lower()]
    offset_cols = [c for c in blinks.columns if "offset" in c.lower()]
    if not onset_cols or not offset_cols:
        return df

    onset_col = onset_cols[0]
    offset_col = offset_cols[0]

    for _, row in blinks.iterrows():
        onset = row[onset_col]
        offset = row[offset_col]
        mask = (df["timestamp"] >= onset) & (df["timestamp"] <= offset)
        df.loc[mask, "is_blink"] = True

    df = df[~df["is_blink"]].reset_index(drop=True)
    return df


# ----------------------------------------------------
# 3. BASHKIMI GAZE + FIXATIONS
# ----------------------------------------------------

def merge_gaze_and_fixations(gaze, fixations):
    g = gaze.copy().sort_values("timestamp").reset_index(drop=True)
    f = fixations.copy().sort_values("start_timestamp").reset_index(drop=True)

    f["fixation_id"] = np.arange(len(f))
    g["fixation_id"] = -1

    for _, fx in f.iterrows():
        mask = (g["timestamp"] >= fx["start_timestamp"]) & (g["timestamp"] <= fx["end_timestamp"])
        g.loc[mask, "fixation_id"] = fx["fixation_id"]

    g = g[g["fixation_id"] >= 0].reset_index(drop=True)
    g = g.merge(f[["fixation_id", "duration_ms"]], on="fixation_id", how="left")
    return g


# ----------------------------------------------------
# 4. PIPELINE PËR NJË SESION
# ----------------------------------------------------

def run_session(data_dir):
    gaze, fixations, blinks, surface_gaze = load_exported_data(data_dir)

    print("Gaze raw shape:", gaze.shape)
    print("Fixations raw shape:", fixations.shape)

    gaze_clean = clean_gaze(gaze)
    fix_clean = clean_fixations(fixations)
    gaze_clean = attach_blinks_to_gaze(gaze_clean, blinks)

    gaze_fix = merge_gaze_and_fixations(gaze_clean, fix_clean)

    print("\nPastruar:")
    print("Gaze:", gaze_clean.shape)
    print("Fixations:", fix_clean.shape)
    print("Merged gaze+fix:", gaze_fix.shape)

    plt.hist(fix_clean["duration_ms"], bins=30)
    plt.xlabel("Fixation duration (ms)")
    plt.ylabel("Count")
    plt.title("Distribuimi i kohëzgjatjes së fiksimeve")
    plt.show()

    print("\nShembuj të të dhënave të bashkuara:")
    print(gaze_fix.head())

    return gaze_fix, fix_clean, surface_gaze


# ----------------------------------------------------
# 5. EKZEKUTIMI
# ----------------------------------------------------

gaze_fix_df, fix_clean_df, surface_gaze_df = run_session(DATA_DIR)
```

```python
# ========= 1. LIDHIM GAZEN ME SURFACE =========

def attach_surface_to_gaze(gaze_fix: pd.DataFrame,
                           surface_gaze: pd.DataFrame,
                           surface_name: str | None = None) -> pd.DataFrame:
    """
    I shton koordinatat e surface-it (x,y në 0–1) te tabela gaze_fix_df.

    Në eksportin e Pupil Labs, surface_gaze_positions.csv zakonisht ka:
      - 'world_index'
      - 'gaze_timestamp' ose 'timestamp'
      - 'norm_pos_x', 'norm_pos_y' (në systemin e surface)
      - 'surface_name' (nëse ka disa surfaces)
    """

    if surface_gaze is None or surface_gaze.empty:
        raise ValueError("surface_gaze është bosh – a e ke aktivizu surface tracking në Pupil Player?")

    s = surface_gaze.copy()

    # filtro vetëm surface-in kryesor nëse ka kolonë surface_name
    name_cols = [c for c in s.columns if "surface_name" in c.lower() or "name" in c.lower()]
    if surface_name is not None and name_cols:
        coln = name_cols[0]
        s = s[s[coln] == surface_name]

    # gjejmë kolonat e koordinatave në surface
    if "norm_pos_x" in s.columns and "norm_pos_y" in s.columns:
        sx, sy = "norm_pos_x", "norm_pos_y"
    else:
        raise ValueError(f"Nuk gjeta norm_pos_x/y te surface_gaze. Kolonat: {list(s.columns)}")

    # bashkojmë sipas world_index (më i thjeshti)
    if "world_index" not in s.columns or "world_index" not in gaze_fix.columns:
        raise ValueError("Nuk gjeta 'world_index' në të dy tabelat për me bo merge.")

    s = s.rename(columns={sx: "surf_x", sy: "surf_y"})

    merged = gaze_fix.merge(
        s[["world_index", "surf_x", "surf_y"]],
        on="world_index",
        how="left"
    )

    # heqim rreshtat ku nuk kemi koordinata surface (jashtë tekstit)
    merged = merged.dropna(subset=["surf_x", "surf_y"]).reset_index(drop=True)

    print("Gaze+fix+surface shape:", merged.shape)
    return merged


# përdorim funksionin për dataset-in tonë aktual
gaze_fix_surf_df = attach_surface_to_gaze(gaze_fix_df, surface_gaze_df)
gaze_fix_surf_df.head()
```
